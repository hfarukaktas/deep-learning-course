```python
# image size is 224x224 as in the original Swin Transformer paper.

IMG_SIZE = 224
random_seed = 42
```

```python
# Caltech101 contains some grayscale data; convert everything to RGB for channel consistency

def to_rgb(img):
    return img.convert("RGB")
```

```python
# data augmentation

train_tf = transforms.Compose([
    transforms.Lambda(to_rgb),
    transforms.RandomAdjustSharpness(sharpness_factor = 2 ,p = 0.1),
    transforms.RandomResizedCrop(IMG_SIZE, scale = (0.6,1), ratio = (0.75,1.33)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]), #imagenet standart values
])

```

```python
test_tf = transforms.Compose([
    transforms.Lambda(to_rgb),
    transforms.Resize(256),
    transforms.CenterCrop(IMG_SIZE),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])
```

```python
from torch.utils.data import random_split

root = "../data"
test_ratio = 0.2

ds_train_full = datasets.Caltech101(root = root, transform = train_tf, download = True)
ds_test_full = datasets.Caltech101(root = root, transform = test_tf, download = True)
```

```python
n_total = len(ds_train_full)
n_test = int(n_total * test_ratio)
n_train = n_total - n_test
```

```python
# train test split

g = torch.Generator().manual_seed(random_seed)
perm = torch.randperm(n_total, generator = g).tolist()

train_idx = perm[:n_train]
test_idx = perm[n_train:]
```

```python
class PatchEmbedding(nn.Module):
    def __init__(self,in_channels:int = 3, embed_dim:int = 96, patch_size: int = 4):
        super().__init__()

        self.patcher = nn.Conv2d(in_channels=in_channels,
                                 out_channels=embed_dim,
                                 kernel_size = patch_size,
                                 stride = patch_size,
                                 padding = 0)

    def forward(self,x):
        x = self.patcher(x)
        return x.permute(0,2,3,1) # (B, 3, 224, 224) -> (B, 96, 56,56) -> (B, 56, 56, 96)
```

```python
class WindowAttention(nn.Module):
    def __init__(self, dim:int = 96, num_heads:int = 3, window_size:int = 7):
        super().__init__()

        self.num_heads = num_heads
        self.qkv = nn.Linear(dim, dim * 3, bias = True) # 96 feature to 3(Q,K,V)x96
        self.proj = nn.Linear(dim, dim)
        self.head_dim = int(dim / num_heads)

        self.scale = self.head_dim ** (-0.5) # 1 / (d square root)

        num_relative_distance = (2 * window_size -1 ) * (2 * window_size - 1) # M = 7 by default, and axis range is (-M+1, M -1)
        self.relative_position_bias_table = nn.Parameter(
            torch.zeros(num_relative_distance, num_heads)
        ) # bias (B) penalty table 169(flattened 13x13)x3(last layer is num_heads channels) for Swin-T
        nn.init.trunc_normal_(self.relative_position_bias_table, std=0.02)

        height_coords = torch.arange(window_size) # 0,1,2,3,4,5,6
        width_coords = torch.arange(window_size)

        coords = torch.stack(torch.meshgrid([height_coords,width_coords], indexing = "ij"))
        # (2,7,7) coords[0] -> all y coords, coords[1] -> all x coords
        coords_flatten = torch.flatten(coords,start_dim = 1, end_dim = 2) # flat the coords matrix to get shape 2x49
        relative_coords = coords_flatten[:,:,None] - coords_flatten[:,None, :] # (2,49,1) - (2,1,49) -> (2,49,49)
        # Channels-Last
        relative_coords = relative_coords.permute(1,2,0).contiguous() # use contiguous() to really change the permutation of the tensor on memory
        # Shifting the [-6,6] range to [0,12] to improve calculation ease by adding (M -1) to every cell
        relative_coords[:,:,0] += (window_size - 1)
        relative_coords[:,:,1] += (window_size - 1)

        relative_coords[:,:,0] *= (2 * window_size) - 1

        relative_position_index = relative_coords.sum(-1)

        self.register_buffer("relative_position_index", relative_position_index)

    def forward(self,x, mask = None):

        B_, N, C = x.shape

        qkv = self.qkv(x).reshape(B_, N,3,self.num_heads,C // self.num_heads).permute(2,0,3,1,4)
        # Q, K, V  needs to come first, number of patches and head dim need to came last

        q,k,v = qkv[0], qkv[1],qkv[2] # separate the Q,K,V values

        # Multihead Attention with Relative Position Bias Formula

        q = q * self.scale # Q / d square root
        attn = q @ k.transpose(-2,-1) # Q * K transposed

        relative_position_bias = self.relative_position_bias_table[self.relative_position_index.view(-1)].view(N,N,-1)

        relative_position_bias = relative_position_bias.permute(2,0,1).contiguous()

        attn = attn + relative_position_bias.unsqueeze(0) # add relative attention bias value

        # in case sw-msa
        if mask is not None:
            nW = mask.shape[0]
            attn = attn.view(B_ // nW, nW, self.num_heads, N, N) + mask.unsqueeze(1).unsqueeze(0)
            attn = attn.view(-1, self.num_heads, N, N)

        attn = torch.softmax(attn, dim=-1) # value goes into softmax nonlinearity before multiply with V
        x = (attn @ v).transpose(1, 2).reshape(B_, N, C)
        x = self.proj(x)
        return x
```

```python
class MLPBlock(nn.Module):
    def __init__(self,num_features:int = 96, hidden_layer: int = 384):
        # 96 * 4 = 384 "the expansion layer of each MLP is α = 4, for all experiments"
        super().__init__()
        self.mlp = nn.Sequential(
            nn.Linear(in_features=num_features, out_features=hidden_layer),
            nn.GELU(),
            nn.Linear(in_features=hidden_layer, out_features=num_features) # after mlp block number of feature goes back to previous value
        )
    def forward(self,x):
        x = self.mlp(x)

        return x
```

```python
def window_partition(x, window_size: int):
    # x: (B, H, W, C) -> (B*nW, M, M, C)
    B, H, W, C = x.shape
    x = x.view(B, H // window_size, window_size, W // window_size, window_size, C)
    windows = x.permute(0, 1, 3, 2, 4, 5).contiguous().view(-1, window_size, window_size, C)
    return windows


def window_reverse(windows, window_size: int, H: int, W: int):
    # windows: (B*nW, M, M, C) -> (B, H, W, C)
    nW = (H // window_size) * (W // window_size)
    B = windows.shape[0] // nW
    x = windows.view(B, H // window_size, W // window_size, window_size, window_size, -1)
    x = x.permute(0, 1, 3, 2, 4, 5).contiguous().view(B, H, W, -1)
    return x
```

```python
def create_attn_mask(H: int, W: int, window_size: int, shift_size: int):
    img_mask = torch.zeros((1, H, W, 1))  # map to keep placement nums

    slices = (slice(0, -window_size),            # first placement
              slice(-window_size, -shift_size),  # shifting size
              slice(-shift_size, None))          # part to add from other edge

    cnt = 0
    for h in slices:
        for w in slices:
            img_mask[:, h, w, :] = cnt
            cnt += 1

    mask_windows = window_partition(img_mask, window_size)              # (nW, M, M, 1)
    mask_windows = mask_windows.view(-1, window_size * window_size)     # (nW, 49)

    attn_mask = mask_windows.unsqueeze(1) - mask_windows.unsqueeze(2)   # (nW, 49, 49)
    attn_mask = attn_mask.masked_fill(attn_mask != 0, -100.0).masked_fill(attn_mask == 0, 0.0)
    return attn_mask
```

```python
class DropPath(nn.Module):
    def __init__(self, drop_prob: float = 0.0):
        super().__init__()
        self.drop_prob = drop_prob

    def forward(self, x):
        if self.drop_prob == 0.0 or not self.training:
            return x
        keep_prob = 1 - self.drop_prob
        shape = (x.shape[0],) + (1,) * (x.ndim - 1)
        random_tensor = keep_prob + torch.rand(shape, dtype=x.dtype, device=x.device)
        random_tensor.floor_()
        return x / keep_prob * random_tensor
```

```python
class SwinTransformerBlock(nn.Module):
    def __init__(self, embedding_dim: int = 96, input_resolution: tuple = (56, 56),
                 num_heads: int = 3, window_size: int = 7, shift_size: int = 0,
                 mlp_hidden: int = None, drop_path: float = 0.0):
        super().__init__()
        self.input_resolution = input_resolution
        self.window_size = window_size
        self.shift_size = shift_size

        if min(input_resolution) <= window_size:
            self.shift_size = 0
            self.window_size = min(input_resolution)

        if mlp_hidden is None:
            mlp_hidden = embedding_dim * 4

        self.norm1 = nn.LayerNorm(normalized_shape=embedding_dim)
        self.norm2 = nn.LayerNorm(normalized_shape=embedding_dim)
        self.wmsa_block = WindowAttention(dim=embedding_dim, num_heads=num_heads,
                                          window_size=self.window_size)
        self.mlp_block = MLPBlock(num_features=embedding_dim, hidden_layer=mlp_hidden)
        self.drop_path = DropPath(drop_path) if drop_path > 0 else nn.Identity()

        if self.shift_size > 0:
            H, W = input_resolution
            attn_mask = create_attn_mask(H, W, self.window_size, self.shift_size)
        else:
            attn_mask = None
        self.register_buffer("attn_mask", attn_mask)

    def forward(self, x):
        H, W = self.input_resolution
        B, L, C = x.shape
        ws = self.window_size

        x_normalized = self.norm1(x).view(B, H, W, C)

        if self.shift_size > 0:
            x_normalized = torch.roll(x_normalized, shifts=(-self.shift_size, -self.shift_size), dims=(1, 2))

        windows = window_partition(x_normalized, ws).view(-1, ws * ws, C)   # (B*nW, 49, C)
        attn_windows = self.wmsa_block(windows, mask=self.attn_mask)        # (B*nW, 49, C)
        attn_windows = attn_windows.view(-1, ws, ws, C)
        x_attn = window_reverse(attn_windows, ws, H, W)                     # (B, H, W, C)

        if self.shift_size > 0:
            x_attn = torch.roll(x_attn, shifts=(self.shift_size, self.shift_size), dims=(1, 2))

        x = x + self.drop_path(x_attn.view(B, H * W, C))

        x_normalized = self.norm2(x)
        x = x + self.drop_path(self.mlp_block(x_normalized))

        return x
```

```python
class PatchMerging(nn.Module):
    def __init__(self, input_resolution: tuple, dim: int):
        super().__init__()
        self.input_resolution = input_resolution
        self.norm = nn.LayerNorm(4 * dim)
        self.reduction = nn.Linear(4 * dim, 2 * dim, bias=False)

    def forward(self, x):
        H, W = self.input_resolution
        B, L, C = x.shape

        x = x.view(B, H, W, C)

        x0 = x[:, 0::2, 0::2, :]
        x1 = x[:, 1::2, 0::2, :]
        x2 = x[:, 0::2, 1::2, :]
        x3 = x[:, 1::2, 1::2, :]

        x = torch.cat([x0, x1, x2, x3], dim=-1)    # (B, H/2, W/2, 4C)
        x = x.view(B, (H // 2) * (W // 2), 4 * C)  # (B, H/2*W/2, 4C)

        x = self.norm(x)
        x = self.reduction(x)                      # (B, H/2*W/2, 2C)
        return x
```

```python
class SwinTransformer(nn.Module):
    def __init__(self, num_classes: int = 101, img_size: int = 224, patch_size: int = 4,
                 embed_dim: int = 96, depths=(2, 2, 6, 2), num_heads=(3, 6, 12, 24),
                 window_size: int = 7, drop_path_rate: float = 0.2):
        super().__init__()

        self.patch_embed = PatchEmbedding(in_channels=3, embed_dim=embed_dim, patch_size=patch_size)
        self.patch_norm = nn.LayerNorm(embed_dim)

        resolution = img_size // patch_size   # 224 / 4 = 56
        dim = embed_dim

        dpr = torch.linspace(0, drop_path_rate, sum(depths)).tolist()
        block_idx = 0

        self.stages = nn.ModuleList()
        for i, (depth, heads) in enumerate(zip(depths, num_heads)):
            layers = []
            for j in range(depth):
                layers.append(SwinTransformerBlock(
                    embedding_dim=dim,
                    input_resolution=(resolution, resolution),
                    num_heads=heads,
                    window_size=window_size,
                    shift_size=0 if j % 2 == 0 else window_size // 2,   # W-MSA, SW-MSA, W-MSA, ...
                    drop_path=dpr[block_idx],
                ))
                block_idx += 1

            if i < len(depths) - 1:
                layers.append(PatchMerging(input_resolution=(resolution, resolution), dim=dim))
                dim *= 2
                resolution //= 2

            self.stages.append(nn.Sequential(*layers))

        self.norm = nn.LayerNorm(dim)
        self.head = nn.Linear(dim, num_classes)

        self.apply(self._init_weights)

    def _init_weights(self, m):
        if isinstance(m, nn.Linear):
            nn.init.trunc_normal_(m.weight, std=0.02)
            if m.bias is not None:
                nn.init.zeros_(m.bias)
        elif isinstance(m, nn.LayerNorm):
            nn.init.ones_(m.weight)
            nn.init.zeros_(m.bias)

    def forward(self, x):
        x = self.patch_embed(x)
        x = x.flatten(1, 2)
        x = self.patch_norm(x)

        for stage in self.stages:
            x = stage(x)

        x = self.norm(x)
        x = x.mean(dim=1)
        return self.head(x)
```

```python
def test_step(model: nn.Module, dataloader: DataLoader, loss_fn: nn.Module, device: str):
    model.eval()
    test_loss, correct, total = 0.0, 0, 0

    with torch.inference_mode():
        for X, y in dataloader:
            X, y = X.to(device), y.to(device)

            logits = model(X)
            loss = loss_fn(logits, y)

            test_loss += loss.item() * X.size(0)
            correct += (logits.argmax(dim=1) == y).sum().item()
            total += y.size(0)

    return test_loss / total, correct / total
```

```python
def train_step(model: nn.Module, dataloader: DataLoader, loss_fn: nn.Module,
               optimizer: torch.optim.Optimizer, device: str,
               scheduler=None, scaler=None, max_grad_norm: float = 5.0):
    model.train()
    train_loss, correct, total = 0.0, 0, 0
    amp_device = "cuda" if device == "cuda" else "cpu"

    for X, y in dataloader:
        X, y = X.to(device), y.to(device)

        optimizer.zero_grad(set_to_none=True)

        with torch.autocast(device_type=amp_device, dtype=torch.float16, enabled=scaler is not None):
            logits = model(X)
            loss = loss_fn(logits, y)

        if scaler is not None:
            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            nn.utils.clip_grad_norm_(model.parameters(), max_grad_norm)
            scaler.step(optimizer)
            scaler.update()
        else:
            loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(), max_grad_norm)
            optimizer.step()

        if scheduler is not None:
            scheduler.step()

        train_loss += loss.item() * X.size(0)
        correct += (logits.argmax(dim=1) == y).sum().item()
        total += y.size(0)

    return train_loss / total, correct / total
```

```python
torch.manual_seed(random_seed)

model = SwinTransformer().to(device)
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4, weight_decay=0.05)
scaler = torch.amp.GradScaler("cuda") if device == "cuda" else None

EPOCHS = 5
results = {"train_loss": [], "train_acc": [], "test_loss": [], "test_acc": []}
best_acc = 0.0
```

```python
for epoch in range(EPOCHS):
    train_loss, train_acc = train_step(model, train_dataloader, loss_fn, optimizer, device, scaler=scaler)
    test_loss, test_acc = test_step(model, test_dataloader, loss_fn, device)

    results["train_loss"].append(train_loss)
    results["train_acc"].append(train_acc)
    results["test_loss"].append(test_loss)
    results["test_acc"].append(test_acc)

    if test_acc > best_acc:
        best_acc = test_acc
        torch.save(model.state_dict(), "swin_t_caltech101_best.pth")

    print(f"Epoch {epoch+1}/{EPOCHS} | train loss {train_loss:.4f} acc {train_acc:.3f} | "
          f"test loss {test_loss:.4f} acc {test_acc:.3f}")

print(f"Best test acc: {best_acc:.3f}")
```
